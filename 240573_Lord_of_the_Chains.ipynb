```python
class Customer:
    def __init__(self, id, name, location, order, restaurant, order_type,d_id):
        self.id = id
        self.name = name
        self.x = location[0]
        self.y = location[1]
        self.order = order
        self.restaurant = restaurant
        self.order_type = order_type
        self.status = "Pending"
```

```python
class Delivery:
    def __init__(self,id,name,location,status,speed):
        self.id=id
        self.name=name
        self.x=location[0]
        self.y=location[1]
        self.status=status
        self.speed = speed
        self.distance = 0
```
